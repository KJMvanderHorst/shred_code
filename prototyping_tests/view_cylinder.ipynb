```python
from pathlib import Path

import numpy as np
import scipy.io as sio
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

from scipy.signal import find_peaks
```

```python
DATA_PATH = Path("../data/cylinder_nektar_wake.mat")

data = sio.loadmat(DATA_PATH)

print("Loaded:", DATA_PATH)
print("Variables in the file:")
print(data.keys())
```

```python
X = data["X_star"]
t = data["t"].squeeze()
U = data["U_star"]
p = data["p_star"]

print("X shape:", X.shape)
print("t shape:", t.shape)
print("U shape:", U.shape)
print("p shape:", p.shape)

print()
print(f"Number of spatial points: {X.shape[0]}")
print(f"Number of time snapshots: {len(t)}")
print(f"Time range: {t[0]:.6f} → {t[-1]:.6f}")
print(f"Total duration: {t[-1] - t[0]:.6f}")
```

```python
x = X[:, 0]
y = X[:, 1]

u = U[:, 0, :]
v = U[:, 1, :]

print("u shape:", u.shape)
print("v shape:", v.shape)
```

```python
velocity_magnitude = np.sqrt(u**2 + v**2)

print("Velocity magnitude shape:", velocity_magnitude.shape)
```

```python
vmin = velocity_magnitude.min()
vmax = velocity_magnitude.max()

print(f"Velocity magnitude range: {vmin:.6f} → {vmax:.6f}")
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

scatter = ax.scatter(
    x,
    y,
    c=velocity_magnitude[:, 0],
    s=8,
    vmin=vmin,
    vmax=vmax
)

ax.set_aspect("equal")
ax.set_xlabel("x")
ax.set_ylabel("y")

title = ax.set_title(
    f"Velocity magnitude, t = {t[0]:.4f}"
)

colorbar = fig.colorbar(scatter, ax=ax)
colorbar.set_label(r"$|\mathbf{u}|$")


def update(frame):
    scatter.set_array(velocity_magnitude[:, frame])

    title.set_text(
        f"Velocity magnitude, "
        f"t = {t[frame]:.4f} "
        f"({frame + 1}/{len(t)})"
    )

    return scatter, title


animation = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=50,
    blit=False
)

plt.close(fig)

display(HTML(animation.to_jshtml()))
```
